# Les résultats des matchs LP

### On commence par regarder combien de matchs sont gagnés à domicile, à l'extérieur ou terminent par un nul.

```python
resultats = pl["FTR"].value_counts()

print(resultats)


```

```python
pourcentage = (pl["FTR"].value_counts(normalize=True) * 100).round(2)
print(pourcentage)
```

### graphe

```python
plt.figure(figsize=(6,4))
plt.bar(pourcentage.index, pourcentage.values,color=["royalblue","gray","darkorange"])

plt.title("Répartition des résultats des matchs")
plt.xlabel("Résultat final")
plt.ylabel("Nombre de matchs")

plt.show()
```

##### Interprétation

##### Si les victoires à domicile (H) sont plus nombreuses que les victoires à l'extérieur (A), cela constitue une première indication d'un avantage du terrain.

##### On observe déjà que les équipes gagnent davantage lorsqu'elles jouent chez elles.

## Les buts marqués

###### On compare ensuite le nombre moyen de buts.

```python
print("Moyenne domicile :", pl["FTHG"].mean())
print("Moyenne extérieur :", pl["FTAG"].mean())
```

```python
nombre_buts_dom = pl["FTHG"].values.sum()
nombre_buts_ext = pl["FTAG"].values.sum()

print("Nombre total de buts marqués à domicile :", nombre_buts_dom)
print("Nombre total de buts marqués à l'extérieur :", nombre_buts_ext)
```

# graphe

```python
moyennes = [pl["FTHG"].mean(), pl["FTAG"].mean()]

plt.figure(figsize=(5,4))
plt.bar(["Domicile","Extérieur"],
        moyennes,
        color=["royalblue","darkorange"])

plt.ylabel("Nombre moyen de buts")
plt.title("Buts moyens marqués")
plt.show()
```

# un test statistique

```python
from scipy.stats import chisquare

resultats = pl["FTR"].value_counts()

test = chisquare(resultats)

print("Statistique Chi² :", test.statistic)
print("p-value :", test.pvalue)
```

# Conclusion pour ton étude

Tu peux écrire :

Le test du Chi² appliqué aux résultats des matchs montre une différence statistiquement significative entre les victoires à domicile, les matchs nuls et les victoires à l'extérieur (χ² = 22,9 ; p < 0,001). Cela indique que les résultats ne sont pas distribués de manière aléatoire et suggère l'existence d'un avantage lié au fait de jouer à domicile.
